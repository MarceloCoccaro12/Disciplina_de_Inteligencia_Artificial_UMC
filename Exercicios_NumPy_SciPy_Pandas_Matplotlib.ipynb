# **2 - Exercícios de SciPy**

```python
# 1 - Calcule a integral da função f(x) = x² sin(x) no intervalo de 0 a π.

from scipy.integrate import quad
import numpy as np

def f(x):
  return x**2 * np.sin(x)

integral = quad(f, 0, np.pi)
print(integral)
```

* print(integral)-> vai imprimir o resultado final na tela, retornando uma tupla com dois valores, (valor_da_integral, estimativa_do_erro_absoluto)

```python
# 2 - Resolva a equação diferencial dy/dx = x − y, com y(0) = 1, no intervalo
# de 0 a 10.
from scipy.integrate import solve_ivp
import numpy as np

def eq_diferencial(t, y):
  return -2 * y

# Resolver a EDO com condicionamento inicla de y(0)=1
solucao = solve_ivp(eq_diferencial, [0,10], [1], t_eval=np.linspace(0,10,100))
print(solucao.y[0])
```

solucao = solve_ivp(...): Executa o solucionador numérico com os seguintes parâmetros:
  * eq_diferencial -> função EDO
  * [0,10] -> intervalo da variável de t=0 até t=10
  * [1] -> condição inicial de y(0)=1
  * t_eval=np.linspace(0, 10, 100): calcular e salvar os valores de y em 100 pontos espaçados entre 0 e 10.



```python
# - 3 Encontre as raízes da função f(x) = x⁴ −3x³+ 2 utilizando SciPy.
from scipy.optimize import root

def f(x):
  return x**4 - 3*x**3 + 2

resultado = root(f, x0=0)
print(resultado.x)
```

```python
# 4 -Use a função de otimização de SciPy para maximizar f(x) = −x²+ 4x+ 1.
from scipy.optimize import minimize

def f(x):
  return -x**2 + 4*x + 1

resultado = minimize(f, x0=0)
print(resultado.fun)
```

* resultado.fun -> Inverte o sinal de volta para obter o y real

```python
# 5 - Calcule a transformada de Fourier de um sinal composto de sin(2πt) e
# cos(4πt).

from scipy.fft import fft
import numpy as np

t= np.linspace(0,1,100)
sinal = np.sin(2*np.pi*t) + np.cos(4*np.pi*t)
transformada = fft(sinal)
print(transformada)
```

```python
# 6 -  Encontre a matriz inversa de uma matriz 5x5 de números aleatórios.
from scipy.linalg import inv
import numpy as np

matriz = np.random.rand(5,5)
matriz_inversa = inv(matriz)
print(matriz)
print()
print(matriz_inversa)
```

```python
# 7 - Use a função de interpolação cúbica de SciPy para interpolar os dados
# (0, 0),(1, 2),(2, 3),(3, 5),(4, 4).
from scipy.interpolate import interp1d
import numpy as np

x = np.array([0,1,2,3,4])
y = np.array([0,2,3,5,4])

interpolacao = interp1d(x,y, kind='cubic')
x_interpolados = np.linspace(0,4,100)
y_interpolados = interpolacao(x_interpolados)

print(x_interpolados)
```

* kind='cubic' -> faz q a interpolação seja cúbica

```python
# 8 - Resolva o sistema de equações 3x + 2y = 5 e x + 4y = 6
from scipy.linalg import solve
import numpy as np

a = np.array([[3,2],[1,4]])
b = np.array([5,6])

solucao = solve(a,b)
print(solucao)
```

```python
# 9 - Auste uma curva quadrática aos pontos (1, 2),(2, 4),(3, 5),(4, 7),(5, 8)
from scipy.optimize import curve_fit
import numpy as np

def curva_quadratica(x, a, b, c):
  return a * x**2 + b * x + c

x = np.array([1,2,3,4,5])
y = np.array([2,4,5,7,8])

parametros, covariância = curve_fit(curva_quadratica, x, y)
print(parametros)

```

```python
# 10 - Calcule a integral dupla da função f(x, y) = xy + x² no intervalo [0, 2]
# para x e [0, 1] para y.

from scipy.integrate import dblquad

f = lambda y, x: x * y + x**2

resultado, erro = dblquad(f, 0, 2, lambda x: 0, lambda x: 1)

print(f"Resultado: {resultado:.4f}")
```

* utilizado a lambda q é uma função que pode ser feita em uma linha só
* A função do dblquad deve receber os parâmetros na ordem (y, x)

```python
# 11- Encontre os autovalores e autovetores de uma matriz 4x4 de números
# inteiros aleatórios

from scipy.linalg import eig
import numpy as np

matriz = np.random.randint(1,100,16).reshape(4,4)
autovalores, autovetores = eig(matriz)
print(matriz)
print()
print(autovalores)
```

* eig -> retorna os autovalores e autovetores da matriz

```python
# 12 - Resolva a equação eˣ −x² = 0 para encontrar a raiz mais próxima de x = 1.
from scipy.optimize import root
import numpy as np

f = lambda x: np.exp(x) - x**2 - 1

resultado = root(f, x0=1)
print(resultado.x)
```

```python
# 13 - Calcule o determinante de uma matriz 6x6 de números aleatórios entre
# -10 e 10.
from scipy.linalg import det
import numpy as np

matriz = np.random.randint(-10,10,36).reshape(6,6)
determinante = det(matriz)
print(matriz)
print()
print(round(determinante))
```

```python
# 14 - Utilize a função de otimização multivariada para minimizar
# f(x, y) = (x−2)² + (y − 3)².
from scipy.optimize import minimize

def f(x):
  return (x[0] - 2)**2 + (x[1] - 3)**2

resultado = minimize(f, x0=[0,0])
print(resultado.x)
```

```python
# 15 - Calcule a transformada de Fourier inversa de um sinal complexo gerado
# aleatoriamente

from scipy.fft import ifft
import numpy as np

sinal = np.random.rand(10) + 1j * np.random.rand(10)
transformada_inversa = ifft(sinal)
print(transformada_inversa)
```

* 1j-> É a unidade equivalente a  i = raiz de -1
* ifft() -> converte os coeficientes do dominio da frequência de volta para o dominio do tempo

```python
# 16 - Resolva a equao diferencial de segunda ordem(d²y/dx²) + y = 0, com condiçõoes
# iniciais y(0) = 1 e y′(0) = 0.
from scipy.integrate import solve_ivp
import numpy as np

def eq_diferencial(t, y):
  return np.array([y[1], -y[0]])

solucao = solve_ivp(eq_diferencial, [0,10], [1,0], t_eval=np.linspace(0,10,100))
print(solucao.y[0])
```

```python
# 18 - Interpole os valores de uma funçãao f(x) = x² nos pontos [1, 2, 3, 4, 5]
# utilizando spline cúbica.
from scipy.interpolate import CubicSpline
import numpy as np

x = np.array([1,2,3,4,5])
y = x**2

interpolacao = CubicSpline(x,y)
x_interpolados = np.linspace(1,5,100)
y_interpolados = interpolacao(x_interpolados)

print(x_interpolados)
```

```python
# 19 - Realize a transformada de Fourier de um sinal discreto composto de senos
# e cossenos.
from scipy.fft import fft
import numpy as np

t = np.linspace(0,1,20)
sinal = np.sin(2*np.pi*t) + np.cos(4*np.pi*t)
transformada = fft(sinal)
print(transformada)
```

```python
# 20 - Resolva um sistema de equaçõeses lineares com três variáveis utilizando SciPy
from scipy.linalg import solve
import numpy as np

#  1x + 2y + 3z = 14
#  2x + 5y + 2z = 18
#  3x + 1y + 5z = 20

# Matriz dos coeficientes (A)
A = np.array([[1,2,3], [2,5,2], [3,1,5]])

# Vetor dos termos independentes (b)
b = np.array([14,18,20])

# Resolução do sistema Ax = b
solucao = solve(A, b)

x, y, z = solucao

print(f"x = {x:.2f}")
print(f"y = {y:.2f}")
print(f"z = {z:.2f}")
```

```python
# 10 - Plote três gráficos de linha em um único gráfico, cada um com uma cor e
# estilo de linha diferentes.
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)

# Plotanso as 3 linhas
plt.plot(x, np.sin(x), "b-", label="Seno")
plt.plot(x, np.cos(x), "r--", label="Cosseno")
plt.plot(x, np.sin(x) + np.cos(x), "g:", label="Soma")
plt.title("Gráfico de Linhas")
plt.xlabel("Eixo X")
plt.ylabel("Eixo Y")
plt.grid(True)
plt.legend()
plt.show()
```

```python
# 11 - Crie um gráfico de barras empilhadas para três categorias com três grupos
# de valores
import matplotlib.pyplot as plt
import numpy as np

categorias = ["Grupo A", "Grupo B", "Grupo C"]

v1 = np.array([10, 15, 8])
v2 = np.array([20, 12, 14])
v3 = np.array([5, 18, 10])

# Criando as barras utilizando o bottom para empilhar
plt.bar(categorias, v1, label="Valores 1", color="blue")
plt.bar(categorias, v2, bottom=v1, label="Valores 2", color="orange")
plt.bar(categorias, v3, bottom=v1 + v2, label="Valores 3", color="green")

plt.title("Gráfico de Barras Empilhadas")
plt.xlabel("\nCategorias")
plt.ylabel("\nValores Acumulados")
plt.legend()

plt.show()
```

* bottom=v1-> Faz com que o segundo grupo de barras comece exatamente acima do topo do primeiro grupo.

* bottom=v1 + v2-> Empilha o terceiro grupo no topo da soma dos dois anteriores.

```python
# 13 - Gere um gráfico de dispersão 3D para 100 pontos aleatórios utilizando
# Matplotlib.
import matplotlib.pyplot as plt
import numpy as np

x = np.random.rand(100)
y = np.random.rand(100)
z = np.random.rand(100)

# Cria a figura e define a projeção 3D
fig = plt.figure()
ax = fig.add_subplot(projection='3d')

ax.scatter(x, y, z)
ax.set_xlabel('Eixo X')
ax.set_ylabel('Eixo Y')
ax.set_zlabel('Eixo Z')
plt.title('Gráfico de Dispersão 3D')
plt.show()


```

* projection='3d' -> Habilita a visualização tridimensional dentro do subplot.
* ax.scatter(x, y, z) -> Pede três vetores de coordenadas para posicionar cada ponto no espaço (x, y, z)
* .ax.set_zlabel() -> É o comando específico para definir o nome do terceiro eixo (Z).
