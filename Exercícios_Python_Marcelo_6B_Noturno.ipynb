```python
# 20 - Criando uma Matriz Transposta, afim de representar a matriz como uma
# lista de listas.
matriz = [[1, 2, 3], [4, 5, 6]]
transposta = []

qntd_linhas = len(matriz)
qntd_colunas = len(matriz[0])

for j in range (qntd_colunas):
  nova_linha = []

  for i in range(qntd_linhas):
    nova_linha.append(matriz[i][j])
  transposta.append(nova_linha)

print("Matriz Transposta:")
for linha in transposta:
  print(linha)

```

```python
# 21 - Implementar a busca binária para encontrar um elemento em uma lista
# ordenada.
lista = [10, 20, 30, 40, 50, 60, 70, 80, 90]
alvo = int(input("Digite um número: "))

inicio = 0
fim = len(lista) - 1
posicao = -1

while (inicio <= fim):

  meio = (inicio + fim) // 2

  if lista[meio] == alvo:
    posicao = meio
    break

  elif lista[meio] < alvo:
    inicio = meio + 1

  else:
    fim = meio - 1

if posicao != -1:
    print(f"O número {alvo} foi encontrado no índice {posicao}.")
else:
    print(f"O número {alvo} não está na lista.")
```

```python
# 30 - Jogo da Forca
# Importar a ferramenta random para pegar palavras aleatórias de uma lista
import random

palavras = ["python", "banana", "jogo", "forca", "computador"]
palavra = random.choice(palavras)
tentativas = 6
letrasCertas = ""

while (tentativas > 0):
  escondido = ""
  for letra in palavra:
    escondido += letra if letra in letrasCertas else " _ "

  print(f"\n Palavra: {escondido} " +"|"+ f" Tentativas: {tentativas}")
  print("-"*75)

  if (escondido == palavra):
    print("Você acertou!")
    break

  chute = input("\n Digite uma letra: ").lower()

  if chute in palavra:
    letrasCertas += chute
  else:
    tentativas -= 1

if (tentativas == 0):
  print("Você perdeu!")

```

```python
# 31 - Implementando o algoritmo de ordenação por bolha (Bubble Sort)
lista = [33,23,56,76,22,11,94]
n = len(lista)

for i in range(n):
  for j in range(0, n-1):
    if lista[j] > lista[j + 1]:
      lista[j], lista[j + 1] = lista[j + 1], lista[j]

print(lista)
```

```python
# 32 - Criando um jogo da velha para dois usuários
board = [" "] * 9
jogador = "X"
jogadas = 0

while jogadas < 9:
    # Desenha o tabuleiro
    print(
        f"\n {board[0]} | {board[1]} | {board[2]} \n---|---|---\n {board[3]} | {board[4]} | {board[5]} \n---|---|---\n {board[6]} | {board[7]} | {board[8]}\n"
    )

    pos = int(input(f"Jogador {jogador}, escolha a posição (1-9): ")) - 1

    if board[pos] != " ":
        print("Posição ocupada! Tente novamente.")
        continue

    board[pos] = jogador
    jogadas += 1

    # Checa todas as combinações de vitória
    vitorias = [
        (0, 1, 2),
        (3, 4, 5),
        (6, 7, 8),  # Linhas
        (0, 3, 6),
        (1, 4, 7),
        (2, 5, 8),  # Colunas
        (0, 4, 8),
        (2, 4, 6),  # Diagonais
    ]

    if any(board[a] == board[b] == board[c] == jogador for a, b, c in vitorias):
        print(f"\nParabéns! O jogador '{jogador}' venceu!")
        break

    jogador = "O" if jogador == "X" else "X"
else:
    print("Empate! Deu velha.")
```

```python
# 37 - Implementando uma função recursiva para calcular o fatorial de um
# número.

def fatorial(n):

  if (n == 0 or n == 1):
    return 1

  return n * fatorial(n -1)

num = int(input("Digite um número: "))
print(f"O fatorial de {num} é {fatorial(num)}")
```

```python
# 40 - Lista de Compras (add, remove listar)
compras = []

while True:
    print("\n--- LISTA DE COMPRAS ---")
    print("1. Adicionar item")
    print("2. Remover item")
    print("3. Listar itens")
    print("4. Sair")

    opcao = int(input("Escolha uma opção: "))

    match opcao:
        case 1:
          item = input("Digite o item a ser adicionado: ")
          compras.append(item)
          print(f"{item} foi adicionado à lista de compras.")

        case 2:
          item = input("Digite o item a ser removido: ")
          if item in compras:
            compras.remove(item)
          else:
            print(f"{item} não está na lista de compras.")

        case 3:
          print("\n Itens na lista de compras:")
          for item in compras:
            print(f"- {item}")

        case 4:
          print("\n Saindo do programa.")
          break

        case _:
          print("\n Opção inválida. Tente novamente.")
```

```python
# 46 - Calculando a média e o desvio padrão de uma lista de números.
import math

numeros = [10, 12, 23, 23, 16, 23, 21, 16]
n = len(numeros)

media = sum(numeros) / n

soma_diferencas_sq = sum((x - media) ** 2 for x in numeros)


desvioPadraoPopu = math.sqrt(soma_diferencas_sq / n)

print(f"Média: {media:.2f}")
print(f"Desvio Padrão Populacional: {desvioPadraoPopu:.2f}")
```

```python
# 48 - Implementando o algoritmo de ordenação por inserção (Insertion Sort).
nums = [12, 11, 13, 5, 6]

for i in range(1, len(nums)):
    chave = nums[i]
    j = i - 1
    while j >= 0 and nums[j] > chave:
        nums[j + 1] = nums[j]
        j -= 1
    nums[j + 1] = chave

print("Ordenado:", nums)
```
